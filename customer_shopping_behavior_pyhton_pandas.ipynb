```python
#checking for null values/missing values
df.isnull().sum()
```

```python
##Imputing missing values in review rating column with median rating of the product category
df['Review Rating' ]=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower().str.replace(r'\s+', '_', regex=True)
```

```python
df.columns
```

```python
#renaming columns in a snake case type for better readability
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(r'\s+', '_', regex=True)
)
```

```python
df.columns
```

```python
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#create new col age_group
labels=['Young adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
df
```

```python
#create purchase_frequency_days
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
```

```python
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
(df['discount_applied']==df['promo_code_used']).all()
```

```python
#droping promocode column coz both colmns are same discount and promo code
df=df.drop('promo_code_used',axis=1)
```
